```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
```

```python
model_metrics = pd.read_csv('model_metrics.csv')
model_metrics.head()
scaler = MinMaxScaler()

nomalized_metrics = scaler.fit_transform(model_metrics[['f1_mean', 'inference_time_mean', 'emissions_mean', 'fpr_mean']])
nomalized_metrics_df = pd.DataFrame(nomalized_metrics, columns=['f1_mean_norm', 'inference_time_mean_norm', 'emissions_mean_norm', 'fpr_mean_norm'])
nomalized_metrics_df.set_index(model_metrics['model'], inplace=True)
nomalized_metrics_df[['inference_time_mean_norm', 'emissions_mean_norm', 'fpr_mean_norm']] = 1 - nomalized_metrics_df[['inference_time_mean_norm', 'emissions_mean_norm', 'fpr_mean_norm']]

nomalized_metrics_df.head()
```
